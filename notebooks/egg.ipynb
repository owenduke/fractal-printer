```python
# Slab

settings = {"power": 2,
 "cx": -1.25,
 "cy": 0.2200000000000002,
 "cz": 0.0,
 "cw": 0.0,
 "slice": 0.0,
 "offset": 0.002,
 "iterations": 10,
 "bailout": 100
}

length = 3
width = 1
thickness = 0.1
julia = fs.quaternion_julia_sdf(**settings)
slab = sdf.d3.box(size = (length, width, thickness), center = (0,0,thickness/2) )

# Generate the points
mesh = mg.generate_mesh(
    julia & slab,
    samples = 2**24,
    simplify = None,
    bounds = mg.box_bounds(size=5),
    save_path = OUTPUT_DIR / "slab.stl"
)
```
